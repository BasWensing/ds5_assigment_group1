```python
# Samengevoegd2

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
 
# belangrijke waardes die niet thuishoren in de functie

x_min = -1.5 # minimale x
x_max = 0.5 # maximale x
y_min = -1  # minimale y
y_max = 1 # maximale y

# figuur tekenen met functie draw_mandel
def draw_mandel(width):
    ''' Definieren van de benodigde lists en benodige x's en y's generen en c's berekenen. Kay '''

    # Stap 1.1 belangrijke lists aanmaken
    c = [] # maakt een lijst aan voor alle c's die we nodig hebben
    x = [] # maakt een lijst voor elke x waarde
    y = [] # maakt een lijst voor elke y waarde
    sequence_mandelbrot = []

    # Stap 1.2 belangrijke x en y's berekenen en elke c uitbereken
    x.extend(np.linspace(x_min, x_max, width)) # Maakt alle benodige x   
    y.extend(np.linspace(y_min, y_max, width)) # Maakt alle benodige y
    
    for x_value in x: # Pakt alle x van de x lijst
        for y_value in y: # Pakt alle y van de y lijst
            c.append(x_value + y_value * 1j) # Rekened de x + y*i uit en voegt deze toe aan een list
    '''Belangrijke parameters toevoegen om verder op te rekenen. Bas '''

    # Stap 1.3 paar parameters nog bepalen voor dat we verder gaan met rekenen            
    n = width # Aantal gegenereerde formulewaarden op interval [x_min, x_max ]
    a0 = 0 # Begin getal
    index = 0 # Index itereren over lijst met functiewaarden
    max_stappen = width+1  # briljant bedacht      

    ''' Met alle c's berekenen en dit in een grid zetten en uiteindelijk plotten. Dennis'''


    # Stap 2: sequence genereren van n getallen in range x = [-1.5, 0.5]
    for c_waarde in c:
        an_oud = 0  # Op 0 beginnen voor elk nieuw complex getal
        n = 0
        # oude waarde checken
        while abs(an_oud) <= 2 and n < max_stappen:
            an_nieuw = an_oud**2 + c_waarde  # Nieuwe waarde berekenen
            an_oud = an_nieuw                # Update an_oud voor volgende stap
            n += 1 # n = n+1
        sequence_mandelbrot.append(abs(an_oud)) 
    # Stap 3: waardes in een grid zetten
    grid = np.array(sequence_mandelbrot).reshape(width,width)# om een figuur aan te maken
    grid_gedraaid = np.rot90(grid) # figuur met 90 graden naar links draaien
    plt.figure(figsize =(8,6)) # bepaalt de grote van de afbeelding
    plt.title('Visualisatie van Mandelbrot (Heatmap)',fontsize=20)  # Geeft een title
    kleuren = LinearSegmentedColormap.from_list( # Geeft de kleuren van de heatmap aan
        "zwart_blauw",
        ["black", "blue"])
    # Stap 4: Alles gaan plotten
    plt.imshow( # functie uit de matplotlib toolbox
            grid_gedraaid, # alle waardes 
            cmap=kleuren, # kleur overgang
            vmin=0, # onderste scheidingsgebeid van de kleuren
            vmax=2, # bovenste scheidingsgebeid van de kleuren
            origin='lower', # zet de y-as goed
            extent=[x_min, x_max, y_min, y_max]
        )

    plt.colorbar(label="Waarde |aₙ|") # Geeft een label aan de heatmap bar
    plt.xlabel('x') # geeft een x een title
    plt.ylabel('y') # geeft een y een title

    plt.tight_layout()  # doet een tight layout
    plt.show() # laat de plot zien

draw_mandel(200)
```

```python
# Exercise 3.1: kleine functie
import random 
import matplotlib.pyplot as plt
import networkx as nx

def klein_netwerk(k):

    # centrum aanmaken
    G = nx.Graph() # G van graph
    centrum = 0
    G.add_node(centrum)

    # links toevoegen
    for i in range(1,k+1):
        G.add_node(i)
        G.add_edge(centrum,i)

    ''' Grafiek van netwerk opslaan in Python. Dennis '''
    return G

''' Voer functie 'G_klein' uit voor bepaalde k (aantal nodes). Dennis '''
G_klein = klein_netwerk(30)

''' Teken netwerk. Dennis '''
nx.draw(G_klein,with_labels=True)

'''Titel plotten. Dennis'''
plt.title('Plot van een klein simpel netwerk')

''' Grafiek visualiseren. Dennis '''
plt.show()
```

```python
# Exercise 3.1: Barbasi-Albert netwerk aanmaken
import random
import matplotlib.pyplot as plt
import networkx as nx

''' Parameters definiëren. Dennis '''


''' Functie 'Barbasi_albert' aanmaken. Dennis '''
def barbasi_albert(N, k, M): 

    ''' Functie 'klein netwerk' uitvoeren voor k punten, grafiek van netwerk opslaan. Dennis '''
    G = klein_netwerk(k)

    ''' Links toevoegen. Bas '''
    for i in range(k + 1, N):

        ''' Bestaande knopen opslaan in lijst, knopen ophalen uit grafiek van netwerk. Dennis '''
        bestaande_knopen = list(G.nodes())

        ''' Di vinden voor de gewichten (Pi). Bas / Dennis '''
        ''' Di = aantal pagina-links dat nummer i bezit. Dennis '''
        ''' Pi = waarschijnlijkheid dat pagina k+i linkt met punt i. Dennis '''

        ''' Graad voor ieder punt i bepaald. Dennis '''
        graden = [G.degree(n) for n in bestaande_knopen]

        ''' Set-functie 'doelen' aanmaken. Dennis '''
        doelen = set()
        
        ''' Juist aantal knopen en links instellen. Bas '''
        while len(doelen) < M: # Als de lengte van 'doelen' kleiner is dan het aantal verschillende pagina's, kunnen er knopen worden toegevoegd. Dennis
            # Knoop kiezen op basis van zijn graad. Bas
            doel = random.choices(bestaande_knopen, weights=graden)[0] # Knoop random kiezen met kans = graden. Dennis
            doelen.add(doel) # Knoop toevoegen aan 'doelen'. Dennis

        ''' Nieuwe verbindingen toevoegen aan knopen in netwerk. Bas / Dennis '''
        for doel in doelen: # Over iedere knoop itereren. Dennis
            G.add_edge(i, doel) # Verbinding (link) toevoegen aan knoop. Dennis

    ''' Netwerktekening opslaan in Python. Dennis '''
    nx.draw(G, with_labels=False)

    ''' Titel en netwerk plotten. Dennis '''
    plt.title(f'Plot van Barabasi-Albert netwerk met N = {N}, k = {k} en M = {M}')
    plt.show()

''' Netwerk aanmaken en opslaan in de variabele G. Bas '''
G = barbasi_albert(10,9,7)

# N = aantal webpagina's. Dennis
# k = aantal knopen rond 1 middelpunt. Dennis
# M = aantal bestaande, verschillende pagina's van het sternetwerk. Dennis

# Netwerk visualiseren
```
